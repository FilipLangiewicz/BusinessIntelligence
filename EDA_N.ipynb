```python
with open("airbnb_data/20241227_australia_sa_barossa-valley_neighbourhoods.geojson", "r", encoding="utf-8") as f:
    data = json.load(f)

features = data.get("features", [])

print(f"Liczba obiektów: {len(features)}\n")

for i, feature in enumerate(features):
    print(f"--- Obiekt #{i + 1} ---")
    
    # Właściwości (opisowe dane)
    properties = feature.get("properties", {})
    print("Właściwości:")
    for key, value in properties.items():
        print(f"  {key}: {value}")

    # Geometria (typ i liczba współrzędnych)
    geometry = feature.get("geometry", {})
    geom_type = geometry.get("type", "Brak")
    coordinates = geometry.get("coordinates", [])

    print(f"Geometria: {geom_type}")
    if geom_type == "Polygon":
        print(f"Liczba punktów (pierwszy pierścień): {len(coordinates[0])}")
    elif geom_type == "MultiPolygon":
        print(f"Liczba wielokątów: {len(coordinates)}")
        if coordinates:
            print(f"Liczba punktów w pierwszym wielokącie: {len(coordinates[0][0])}")
    print()
```
